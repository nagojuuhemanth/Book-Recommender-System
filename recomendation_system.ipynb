```python
selected_columns = [
    "bookId",
    "title",
    "author",
    "description",
    "genres",
    "rating",
    "coverImg",
    "language"
]

books = df[selected_columns].copy()

print("Before cleaning:")
print(books.shape)

print("\nMissing values:")
print(books.isnull().sum())
```

```python
# Remove duplicate book IDs
books = books.drop_duplicates(
    subset="bookId",
    keep="first"
).reset_index(drop=True)

print("Dataset shape:", books.shape)
print("Duplicate bookIds:", books["bookId"].duplicated().sum())
```

```python
print("Missing values:")
print(books.isnull().sum())

print("\nDuplicate bookIds:",
      books["bookId"].duplicated().sum())

print("\nEmpty titles:",
      books["title"].str.strip().eq("").sum())

print("Empty authors:",
      books["author"].str.strip().eq("").sum())

print("Empty genres:",
      books["genres"].str.strip().eq("").sum())

print("Empty descriptions:",
      books["description"].str.strip().eq("").sum())
```

```python
import pandas as pd

books = pd.read_csv("/processed_data/books_processed.csv")

print(books.shape)
print(books.columns.tolist())
```

```python
nlp_books = books[
    [
        "bookId",
        "title",
        "author",
        "genres",
        "description"
    ]
].copy()
```

```python
nlp_books.head()
```

```python
nlp_books.shape
```

```python
import ast

def clean_genres(genre_string):
    try:
        genres = ast.literal_eval(genre_string)
        return ", ".join(genres)
    except:
        return str(genre_string)


nlp_books["genres_clean"] = nlp_books["genres"].apply(clean_genres)
```

```python
display(
    nlp_books[
        ["genres", "genres_clean"]
    ].head(10)
)
```

```python
nlp_books["text"] = (
    "Title: " + nlp_books["title"].astype(str) +
    ". Author: " + nlp_books["author"].astype(str) +
    ". Genres: " + nlp_books["genres_clean"].astype(str) +
    ". Description: " + nlp_books["description"].astype(str)
)
```

```python
print(nlp_books["text"].iloc[0])
```

```python
for i in range(3):
    print(f"\n--- Book {i+1} ---")
    print(nlp_books["text"].iloc[i])
```

```python
display(
    nlp_books[
        ["title", "genres_clean", "description", "text"]
    ].head(3)
)
```

```python
print(nlp_books["text"].str.len().describe())
```

```python
# Make sure text is string
nlp_books["text"] = nlp_books["text"].astype(str)

# Replace newlines, tabs and repeated whitespace
nlp_books["text"] = (
    nlp_books["text"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
print("Empty text:",
      nlp_books["text"].str.len().eq(0).sum())

print("Missing text:",
      nlp_books["text"].isnull().sum())
```

```python
for i in range(3):
    print(f"\n--- Book {i+1} ---")
    print(nlp_books["text"].iloc[i])
```

```python
nlp_final = nlp_books[
    [
        "bookId",
        "title",
        "author",
        "genres",
        "description",
        "text"
    ]
].copy()
```

```python
print("Shape:", nlp_final.shape)
print("\nColumns:")
print(nlp_final.columns.tolist())
```
